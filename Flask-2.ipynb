Q1. Explain GET and POST methods.

Ans:- GET and POST are two commonly used HTTP methods in web applications, GET Method: Used to request data from the server, The parameters are sent in the URL, making it suitable for retrieving information but not secure for sensitive data, Example: searching on Google, POST Method: Used to send data to the server, usually in the body of the request, It is more secure and is commonly used for submitting forms, login details, or uploading files, Example: submitting a registration form.

Q2. Why is request used in Flask?

Ans:- In Flask, the request object is used to access data sent by the client to the server, It allows us to retrieve form data, query parameters, JSON payloads, and headers, For example, when a user submits a form, Flask uses request.form to capture the input values. Without request, the server would not be able to process client inputs.

Q3. Why is redirect() used in Flask?

Ans:- The redirect() function in Flask is used to send the user to a different URL, It is helpful when we want to move the user from one route to another after performing an action, For example, after a successful login, we can redirect the user to the dashboard page, This improves navigation and user experience.

Q4. What are templates in Flask? Why is the render_template() function used?

Ans:- Templates in Flask are HTML files that allow dynamic content rendering, They are usually stored in a folder named templates, Flask uses the Jinja2 template engine to insert Python variables and logic into HTML, The render_template() function is used to load and render these HTML files, It allows us to pass data from Python code to the template, making web pages dynamic and interactive.

Q5. Create a simple API. Use Postman to test it. Attach the screenshot of the output in the Jupyter Notebook.

```python
from flask import Flask, jsonify, request

app = Flask(__name__)

# Simple API route
@app.route('/api/data', methods=['GET', 'POST'])
def api_data():
    if request.method == 'GET':
        return jsonify({"message": "This is a GET request", "status": "success"})
    elif request.method == 'POST':
        data = request.json
        return jsonify({"message": "POST request received", "data": data, "status": "success"})

if __name__ == "__main__":
    app.run(debug=True)
```
